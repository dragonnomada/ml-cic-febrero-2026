### Análisis cruzado / combinado

* num-num: Puntos
* num-cat: Cajas segmentadas
* cat-num: Densidad segmentada
* cat-cat: Mapa de calor

```python
seaborn.scatterplot(x=c4, y=c7)
```

```python
seaborn.boxplot(y=c4, hue=c1, palette="Set1")
```

```python
seaborn.kdeplot(x=c7, hue=c1)
```

```python
pandas.DataFrame([c2, c9]).T.value_counts().unstack()
```

```python
r1 = pandas.DataFrame([c2, c9]).T.value_counts().unstack()

r1 = 100 * r1.div(r1.sum(axis=0), axis=1).T

r1
```

```python
r2 = pandas.DataFrame([c2, c9]).T.value_counts().unstack()

r2 = 100 * r2.div(r2.sum(axis=1), axis=0)

r2
```

```python
seaborn.heatmap(r1, cmap="BuPu", annot=True)
```

```python
seaborn.heatmap(r2, cmap="BuPu", annot=True)
```
