```python
def lista_n_primos(n = 20):
    primos = []
    i = 2
    while len(primos) < n:
        esPrimo = True
        for primo in primos:
            if i % primo == 0:
                esPrimo = False
        if esPrimo:
            primos.append(i)
        i += 1
    return primos
```

```python
lista_n_primos()
```

```python
lista_n_primos(5)
```

```python
def transformarGenero(genero):
    if genero.upper()[0] == "H":
        return -1
    elif genero.upper()[0] == "M":
        return 1
    else:
        return 0
```

```python
transformarGenero("hombre")
```

```python
transformarGenero("h")
```

```python
transformarGenero("Hombre")
```

```python
transformarGenero("Mujer")
```
