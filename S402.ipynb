```python
x15.name = "Baths"

x0 = pandas.Series(numpy.ones(len(y)), name="bias")

X = pandas.DataFrame([x0, x2, x4, x15, x7, x8, x9, x10, x11, x12, x13, x14]).T # k=12

X
```

$$
y = \beta_0 + \beta_1 * x_1 + ... + \beta_k * x_k
$$

$$
y = X \cdot \beta
$$

$$
\beta = (X^T \cdot X)^{-1} X^T y
$$

```python
X.shape
```

```python
259/1460
```

```python
X.isna().sum()
```

```python
beta = numpy.linalg.inv(X.T.dot(X)).dot(X.T).dot(y) # x0, x2, x4, x15, x7, x8, x9, x10, x11, x12, x13, x14

pandas.DataFrame(beta, index=["x0", "x2", "x4", "x15", "x7", "x8", "x9", "x10", "x11", "x12", "x13", "x14"])
```

```python
X.describe()
```

```python
1973*193.7
```

```python
yp = X.dot(beta)

pandas.DataFrame([y, yp]).T
```

$$
MSE = (y - \hat{y}) \cdot (y - \hat{y})T
$$

```python
seaborn.scatterplot(x=y, y=yp, marker=".")
```

```python
e = y - yp

seaborn.kdeplot(x=e)
```

```python
# MSE

MSE = (y - yp).dot((y - yp).T) / len(y)

# RMSE

RMSE = MSE ** 0.5

# MAE

MAE = (y - yp).abs().sum() / len(y)

RMSE, MAE
```

```python
X = pandas.DataFrame([x0, x2, x4, x15, x7, x8, x9, x10, x11, x12, x13, x14]).T # k=7

X
```

```python
from sklearn.linear_model import LinearRegression

modelo = LinearRegression()

modelo.fit(X, y)

modelo.coef_, modelo.intercept_
```

```python
yp = modelo.predict(X)

seaborn.scatterplot(x=y, y=yp, marker=".")
```

```python
e = y - yp

seaborn.kdeplot(x=e)
```

```python
# MSE

MSE = (y - yp).dot((y - yp).T) / len(y)

# RMSE

RMSE = MSE ** 0.5

# MAE

MAE = (y - yp).abs().sum() / len(y)

RMSE, MAE
```
