```python
import pandas
import seaborn
```

```python
titanic = pandas.read_csv("titanic.csv")

titanic.head()
```

```python
# numérica
x1 = titanic["Age"]

x1.plot.box(vert=False)
```

```python
x1.plot.density()
```

```python
x2 = titanic["Sex"]

x2.value_counts()
```

```python
x2.value_counts(normalize=True)
```

```python
x2.value_counts().plot.bar()
```

```python
x2.value_counts().plot.pie(autopct="%.2f%%")
```

```python
seaborn.boxplot(y=x1, hue=x2)
```

```python
seaborn.violinplot(y=x1, hue=x2)
```

```python
seaborn.kdeplot(x=x1, hue=x2)
```

```python
x3 = titanic["Survived"]

x3.value_counts()
```

```python
seaborn.kdeplot(x=x1, hue=x3)
```
